```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
from urllib.parse import quote_plus
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 120

# ── Put your credentials here ──
DB_USER     = "postgres"
DB_PASSWORD = quote_plus("Harshit@123")  # quote_plus handles @, #, $ safely
DB_HOST     = "localhost"
DB_PORT     = "5432"
DB_NAME     = "credit_risk_db"

engine = create_engine(
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

# Test connection
with engine.connect() as conn:
    print("✓ Connected successfully to credit_risk_db")
```

```python
customers    = pd.read_sql("SELECT * FROM customers",    engine)
cards        = pd.read_sql("SELECT * FROM cards",        engine)
merchants    = pd.read_sql("SELECT * FROM merchants",    engine)
transactions = pd.read_sql("SELECT * FROM transactions", engine)

print(f"customers    : {len(customers):,} rows")
print(f"cards        : {len(cards):,} rows")
print(f"merchants    : {len(merchants):,} rows")
print(f"transactions : {len(transactions):,} rows")
```

```python
# Join all 4 tables into one master DataFrame
master = (
    transactions
    .merge(cards,      on='card_id',      how='left')
    .merge(customers,  on='customer_id',  how='left')
    .merge(merchants,  on='merchant_id',  how='left')
)

master['txn_time']    = pd.to_datetime(master['txn_time'])
master['signup_date'] = pd.to_datetime(master['signup_date'])
master['issue_date']  = pd.to_datetime(master['issue_date'])

print(f"Master shape : {master.shape}")
print(f"Columns      : {list(master.columns)}")
```

```python
print("=== DATA TYPES ===")
print(master.dtypes)

print("\n=== NULL CHECK ===")
print(master.isnull().sum())

print("\n=== TRANSACTION STATUS SPLIT ===")
print(master['status'].value_counts(normalize=True).mul(100).round(2).astype(str) + '%')

print("\n=== AMOUNT STATS ===")
print(master['amount'].describe().round(2))

print("\n=== DATE RANGE ===")
print(f"Earliest txn : {master['txn_time'].min()}")
print(f"Latest txn   : {master['txn_time'].max()}")
```

```python
snapshot_date = master['txn_time'].max()

# ── APPROVED transactions only for spend features
approved = master[master['status'] == 'approved'].copy()

# ── Per-customer aggregations
customer_features = approved.groupby('customer_id').agg(
    total_spend       = ('amount', 'sum'),
    avg_ticket_size   = ('amount', 'mean'),
    txn_count         = ('amount', 'count'),
    last_txn_date     = ('txn_time', 'max'),
    first_txn_date    = ('txn_time', 'min'),
    merchant_diversity= ('category', 'nunique'),
).reset_index()

# ── RFM Features
customer_features['recency_days']  = (snapshot_date - customer_features['last_txn_date']).dt.days
customer_features['frequency']     = customer_features['txn_count']
customer_features['monetary']      = customer_features['total_spend']

# ── Spend velocity (spend per active day)
customer_features['active_days']   = (
    customer_features['last_txn_date'] - customer_features['first_txn_date']
).dt.days.replace(0, 1)
customer_features['spend_velocity'] = (
    customer_features['total_spend'] / customer_features['active_days']
).round(2)

print("✓ Core features built")
print(customer_features.head(3))
```

```python
# ── Decline & dispute rates (all transactions, not just approved)
status_features = master.groupby('customer_id').agg(
    total_txns   = ('transaction_id', 'count'),
    declined     = ('status', lambda x: (x == 'declined').sum()),
    disputed     = ('status', lambda x: (x == 'disputed').sum()),
).reset_index()

status_features['decline_rate']  = (status_features['declined'] / status_features['total_txns']).round(4)
status_features['dispute_rate']  = (status_features['disputed'] / status_features['total_txns']).round(4)

# ── Late night ratio (11pm to 4am)
master['hour'] = master['txn_time'].dt.hour
master['is_late_night'] = master['hour'].between(23, 23) | master['hour'].between(0, 4)

late_night = master.groupby('customer_id').agg(
    late_night_txns = ('is_late_night', 'sum'),
    total_txns_ln   = ('transaction_id', 'count')
).reset_index()
late_night['late_night_ratio'] = (late_night['late_night_txns'] / late_night['total_txns_ln']).round(4)

print("✓ Decline, dispute, late-night features built")
```

```python
# ── Average spend per card vs credit limit
card_spend = approved.groupby('card_id')['amount'].sum().reset_index()
card_spend.columns = ['card_id', 'card_total_spend']

card_util = cards.merge(card_spend, on='card_id', how='left').fillna(0)
card_util['utilisation'] = (card_util['card_total_spend'] / card_util['credit_limit']).round(4)

# Average utilisation per customer (they may have multiple cards)
util_by_customer = card_util.groupby('customer_id')['utilisation'].mean().reset_index()
util_by_customer.columns = ['customer_id', 'avg_utilisation']

print("✓ Credit utilisation feature built")
```

```python
# ── Merge everything together
features = (
    customer_features
    .merge(status_features[['customer_id','decline_rate','dispute_rate']], on='customer_id', how='left')
    .merge(late_night[['customer_id','late_night_ratio']],                 on='customer_id', how='left')
    .merge(util_by_customer,                                               on='customer_id', how='left')
    .merge(customers[['customer_id','income_band','city','age']],          on='customer_id', how='left')
)

# ── RFM Score (1–5 scale, 5 = best)
features['R_score'] = pd.qcut(features['recency_days'],  q=5, labels=[5,4,3,2,1]).astype(int)
features['F_score'] = pd.qcut(features['frequency'].rank(method='first'), q=5, labels=[1,2,3,4,5]).astype(int)
features['M_score'] = pd.qcut(features['monetary'].rank(method='first'),  q=5, labels=[1,2,3,4,5]).astype(int)
features['RFM_score'] = features['R_score'] + features['F_score'] + features['M_score']

# ── RFM Segment labels
def rfm_segment(score):
    if score >= 13: return 'Champions'
    elif score >= 10: return 'Loyal'
    elif score >= 7: return 'At Risk'
    else: return 'Lost'

features['rfm_segment'] = features['RFM_score'].apply(rfm_segment)

print(f"✓ Final feature set: {features.shape}")
print(features['rfm_segment'].value_counts())
```

```python
features.to_csv('customer_features.csv', index=False)
print("✓ customer_features.csv saved")
print(f"\nFinal columns:\n{list(features.columns)}")
```
