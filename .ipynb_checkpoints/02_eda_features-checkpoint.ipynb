```python
import os

os.makedirs("data", exist_ok=True)
os.makedirs("charts", exist_ok=True)
os.makedirs("reports", exist_ok=True)

print("✓ All folders created")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 120
plt.rcParams['figure.facecolor'] = 'white'

# Load all tables
customers    = pd.read_csv('data/customers.csv')
sellers      = pd.read_csv('data/sellers.csv')
products     = pd.read_csv('data/products.csv')
orders       = pd.read_csv('data/orders.csv')
order_items  = pd.read_csv('data/order_items.csv')
returns      = pd.read_csv('data/returns.csv')

# Parse dates
orders['order_date'] = pd.to_datetime(orders['order_date'])
returns['return_date'] = pd.to_datetime(returns['return_date'])

print("All tables loaded")
for name, df in zip(
    ['customers','sellers','products','orders','order_items','returns'],
    [customers, sellers, products, orders, order_items, returns]
):
    print(f"  {name:<15}: {len(df):>8,} rows")
```

```python
# Join all tables into one master analytical DataFrame
master = (
    order_items
    .merge(orders[['order_id','customer_id','order_date','payment_method',
                   'delivery_delay_days','order_value']], on='order_id', how='left')
    .merge(customers[['customer_id','city','state','customer_segment',
                      'account_age_days']], on='customer_id', how='left')
    .merge(products[['product_id','category','subcategory','price',
                     'avg_rating','image_count']], on='product_id', how='left')
    .merge(sellers[['seller_id','rating','fulfilment_type']],
           on='seller_id', how='left')
)

master['order_month'] = master['order_date'].dt.month
master['order_quarter'] = master['order_date'].dt.quarter
master['is_cod'] = (master['payment_method'] == 'COD').astype(int)
master['is_flipkart_fulfilled'] = (master['fulfilment_type'] == 'Flipkart Fulfilled').astype(int)

print(f"Master shape: {master.shape}")
print(f"\nNull check:")
print(master.isnull().sum()[master.isnull().sum() > 0])
print(f"\nOverall return rate: {master['is_returned'].mean()*100:.2f}%")
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

cat_returns = master.groupby('category').agg(
    total_items=('is_returned', 'count'),
    returned=('is_returned', 'sum')
).reset_index()
cat_returns['return_rate'] = (cat_returns['returned'] / cat_returns['total_items'] * 100).round(2)
cat_returns = cat_returns.sort_values('return_rate', ascending=False)

colors = ['#E24B4A' if r > 25 else '#EF9F27' if r > 15 else '#1D9E75'
          for r in cat_returns['return_rate']]

bars = ax.bar(cat_returns['category'], cat_returns['return_rate'],
              color=colors, edgecolor='none', width=0.6)

for bar, val in zip(bars, cat_returns['return_rate']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.3,
            f'{val}%', ha='center', fontsize=9, fontweight='bold')

ax.axhline(master['is_returned'].mean()*100, color='navy',
           linestyle='--', linewidth=1.5, label=f"Overall avg ({master['is_returned'].mean()*100:.1f}%)")
ax.set_title('Return Rate by Product Category', fontsize=13, fontweight='bold')
ax.set_xlabel('Category')
ax.set_ylabel('Return Rate (%)')
ax.legend()
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('charts/chart1_return_by_category.png', bbox_inches='tight')
plt.show()
print("✓ Chart 1 saved")
```

```python
import os
os.makedirs('charts', exist_ok=True)

fig, ax = plt.subplots(figsize=(10, 5))

pay_returns = master.groupby('payment_method').agg(
    total=('is_returned', 'count'),
    returned=('is_returned', 'sum')
).reset_index()
pay_returns['return_rate'] = (pay_returns['returned'] / pay_returns['total'] * 100).round(2)
pay_returns = pay_returns.sort_values('return_rate', ascending=False)

bars = ax.bar(pay_returns['payment_method'], pay_returns['return_rate'],
              color=['#E24B4A' if m == 'COD' else '#1E3A8A'
                     for m in pay_returns['payment_method']],
              edgecolor='none', width=0.5)

for bar, val in zip(bars, pay_returns['return_rate']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.2,
            f'{val}%', ha='center', fontsize=9, fontweight='bold')

ax.set_title('Return Rate by Payment Method\n(COD highlighted in red — highest return risk)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Payment Method')
ax.set_ylabel('Return Rate (%)')
plt.tight_layout()
plt.savefig('charts/chart2_return_by_payment.png', bbox_inches='tight')
plt.show()
print("✓ Chart 2 saved")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

master['delay_bucket'] = pd.cut(
    master['delivery_delay_days'],
    bins=[-1, 0, 1, 2, 3, 5, 100],
    labels=['On time', '1 day', '2 days', '3 days', '4-5 days', '5+ days']
)

delay_returns = master.groupby('delay_bucket', observed=True).agg(
    total=('is_returned', 'count'),
    returned=('is_returned', 'sum')
).reset_index()
delay_returns['return_rate'] = (delay_returns['returned'] / delay_returns['total'] * 100).round(2)

ax.plot(delay_returns['delay_bucket'].astype(str),
        delay_returns['return_rate'],
        marker='o', linewidth=2.5, color='#E24B4A', markersize=8)

for i, row in delay_returns.iterrows():
    ax.annotate(f"{row['return_rate']}%",
                (str(row['delay_bucket']), row['return_rate']),
                textcoords='offset points', xytext=(0, 10),
                ha='center', fontsize=9, fontweight='bold')

ax.fill_between(range(len(delay_returns)), delay_returns['return_rate'],
                alpha=0.1, color='#E24B4A')
ax.set_title('Return Rate by Delivery Delay\n(Key insight: every extra day of delay raises return risk)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Delivery Delay')
ax.set_ylabel('Return Rate (%)')
ax.set_xticks(range(len(delay_returns)))
ax.set_xticklabels(delay_returns['delay_bucket'].astype(str))
plt.tight_layout()
plt.savefig('charts/chart3_return_by_delay.png', bbox_inches='tight')
plt.show()
print("✓ Chart 3 saved")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

master['discount_bucket'] = pd.cut(
    master['discount_pct'],
    bins=[-1, 0, 10, 20, 30, 40, 100],
    labels=['No discount', '1-10%', '11-20%', '21-30%', '31-40%', '40%+']
)

disc_returns = master.groupby('discount_bucket', observed=True).agg(
    total=('is_returned', 'count'),
    returned=('is_returned', 'sum')
).reset_index()
disc_returns['return_rate'] = (disc_returns['returned'] / disc_returns['total'] * 100).round(2)

bars = ax.bar(disc_returns['discount_bucket'].astype(str),
              disc_returns['return_rate'],
              color=['#1D9E75','#1D9E75','#EF9F27','#EF9F27','#E24B4A','#E24B4A'],
              edgecolor='none', width=0.6)

for bar, val in zip(bars, disc_returns['return_rate']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.2,
            f'{val}%', ha='center', fontsize=9, fontweight='bold')

ax.set_title('Return Rate by Discount Level\n(Heavy discounts drive impulse buys — and more returns)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Discount Level')
ax.set_ylabel('Return Rate (%)')
plt.tight_layout()
plt.savefig('charts/chart4_return_by_discount.png', bbox_inches='tight')
plt.show()
print("✓ Chart 4 saved")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

master['seller_rating_bucket'] = pd.cut(
    master['rating'],
    bins=[0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.1],
    labels=['< 2.5', '2.5-3.0', '3.0-3.5', '3.5-4.0', '4.0-4.5', '4.5+']
)

seller_returns = master.groupby('seller_rating_bucket', observed=True).agg(
    total=('is_returned', 'count'),
    returned=('is_returned', 'sum')
).reset_index()
seller_returns['return_rate'] = (seller_returns['returned'] / seller_returns['total'] * 100).round(2)

ax.plot(seller_returns['seller_rating_bucket'].astype(str),
        seller_returns['return_rate'],
        marker='o', linewidth=2.5, color='#6B21A8', markersize=8)

for i, row in seller_returns.iterrows():
    ax.annotate(f"{row['return_rate']}%",
                (str(row['seller_rating_bucket']), row['return_rate']),
                textcoords='offset points', xytext=(0, 10),
                ha='center', fontsize=9, fontweight='bold')

ax.fill_between(range(len(seller_returns)), seller_returns['return_rate'],
                alpha=0.1, color='#6B21A8')
ax.set_title('Return Rate by Seller Rating\n(Low-rated sellers drive disproportionate returns)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Seller Rating')
ax.set_ylabel('Return Rate (%)')
ax.set_xticks(range(len(seller_returns)))
ax.set_xticklabels(seller_returns['seller_rating_bucket'].astype(str))
plt.tight_layout()
plt.savefig('charts/chart5_return_by_seller_rating.png', bbox_inches='tight')
plt.show()
print("✓ Chart 5 saved")
```

```python
print("Engineering features for ML model...")

# ── Customer-level features ────────────────────────────────
cust_features = master.groupby('customer_id').agg(
    cust_total_orders=('order_id', 'nunique'),
    cust_total_items=('item_id', 'count'),
    cust_return_rate=('is_returned', 'mean'),
    cust_avg_order_value=('order_value', 'mean'),
    cust_cod_ratio=('is_cod', 'mean'),
).reset_index()

# ── Seller-level features ──────────────────────────────────
seller_features = master.groupby('seller_id').agg(
    seller_total_items=('item_id', 'count'),
    seller_return_rate=('is_returned', 'mean'),
    seller_avg_price=('unit_price', 'mean'),
    seller_avg_discount=('discount_pct', 'mean'),
).reset_index()
seller_features.columns = ['seller_id', 'seller_total_items',
                           'seller_return_rate', 'seller_avg_price',
                           'seller_avg_discount']

# ── Product-level features ─────────────────────────────────
prod_features = master.groupby('product_id').agg(
    prod_total_sold=('item_id', 'count'),
    prod_return_rate=('is_returned', 'mean'),
    prod_avg_discount_given=('discount_pct', 'mean'),
).reset_index()

# ── Merge all back into master ─────────────────────────────
features = (
    master
    .merge(cust_features, on='customer_id', how='left')
    .merge(seller_features, on='seller_id', how='left')
    .merge(prod_features, on='product_id', how='left')
)

# ── Boolean / derived features ────────────────────────────
features['is_high_discount'] = (features['discount_pct'] >= 40).astype(int)
features['is_delayed'] = (features['delivery_delay_days'] > 0).astype(int)
features['is_new_customer'] = (features['account_age_days'] < 30).astype(int)
features['is_low_rated_seller'] = (features['rating'] < 3.5).astype(int)
features['is_low_image_count'] = (features['image_count'] <= 2).astype(int)
features['is_high_price'] = (features['unit_price'] > 20000).astype(int)
features['price_bucket'] = pd.qcut(features['unit_price'], q=5, labels=[1,2,3,4,5]).astype(int)

print(f"✓ Feature set shape: {features.shape}")
print(f"\nFeature columns engineered:")
new_cols = ['cust_total_orders','cust_return_rate','cust_cod_ratio',
            'seller_return_rate','prod_return_rate',
            'is_high_discount','is_delayed','is_new_customer',
            'is_low_rated_seller','is_low_image_count','is_high_price']
for c in new_cols:
    print(f"  {c}")
```

```python
feature_cols = [
    'is_returned', 'unit_price', 'discount_pct', 'delivery_delay_days',
    'rating', 'avg_rating', 'image_count', 'account_age_days',
    'is_cod', 'is_flipkart_fulfilled', 'is_high_discount',
    'is_delayed', 'is_new_customer', 'is_low_rated_seller',
    'cust_return_rate', 'seller_return_rate', 'prod_return_rate'
]

corr = features[feature_cols].corr().round(2)

fig, ax = plt.subplots(figsize=(14, 10))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f',
            cmap='RdYlGn', center=0, ax=ax,
            linewidths=0.5, cbar_kws={'label': 'Correlation'})
ax.set_title('Feature Correlation Heatmap', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('charts/chart6_correlation_heatmap.png', bbox_inches='tight')
plt.show()
print("✓ Chart 6 saved")
```

```python
features.to_csv('data/features.csv', index=False)

print("✓ features.csv saved")
print(f"Shape: {features.shape}")
print(f"\nReturn rate in feature set: {features['is_returned'].mean()*100:.2f}%")
print(f"\nCharts saved:")
import os
for f in sorted(os.listdir('charts')):
    print(f"  {f}")
```
