```python
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random
import os
import warnings
warnings.filterwarnings('ignore')

fake = Faker('en_IN')
np.random.seed(42)
random.seed(42)

os.makedirs('data', exist_ok=True)
print("✓ Libraries ready")
```

```python
CITIES = ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad', 'Chennai',
          'Pune', 'Kolkata', 'Ahmedabad', 'Jaipur', 'Surat',
          'Lucknow', 'Kanpur', 'Nagpur', 'Indore', 'Bhopal']

STATES = {
    'Mumbai': 'Maharashtra', 'Pune': 'Maharashtra', 'Nagpur': 'Maharashtra',
    'Delhi': 'Delhi', 'Jaipur': 'Rajasthan',
    'Bangalore': 'Karnataka', 'Hyderabad': 'Telangana',
    'Chennai': 'Tamil Nadu', 'Kolkata': 'West Bengal',
    'Ahmedabad': 'Gujarat', 'Surat': 'Gujarat',
    'Lucknow': 'Uttar Pradesh', 'Kanpur': 'Uttar Pradesh',
    'Indore': 'Madhya Pradesh', 'Bhopal': 'Madhya Pradesh'
}

N_CUSTOMERS = 50000

segments = np.random.choice(
    ['Regular', 'Premium', 'New'],
    size=N_CUSTOMERS,
    p=[0.60, 0.25, 0.15]
)

account_age = np.where(
    segments == 'New',
    np.random.randint(1, 30, N_CUSTOMERS),
    np.where(
        segments == 'Premium',
        np.random.randint(365, 1825, N_CUSTOMERS),
        np.random.randint(30, 1000, N_CUSTOMERS)
    )
)

cities = np.random.choice(CITIES, size=N_CUSTOMERS)

customers = pd.DataFrame({
    'customer_id': [f'CUST{str(i).zfill(6)}' for i in range(1, N_CUSTOMERS+1)],
    'name': [fake.name() for _ in range(N_CUSTOMERS)],
    'city': cities,
    'state': [STATES[c] for c in cities],
    'customer_segment': segments,
    'account_age_days': account_age,
    'signup_date': [
        (datetime(2024,1,1) - timedelta(days=int(d))).strftime('%Y-%m-%d')
        for d in account_age
    ]
})

customers.to_csv('data/customers.csv', index=False)
print(f"✓ Customers: {len(customers):,} rows")
print(customers['customer_segment'].value_counts())
```

```python
N_SELLERS = 5000

SELLER_CITIES = ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad',
                 'Chennai', 'Pune', 'Kolkata', 'Surat', 'Ahmedabad']

# Sellers have embedded return-rate signal via their rating
seller_ratings = np.round(np.clip(np.random.normal(3.8, 0.7, N_SELLERS), 1.0, 5.0), 1)

sellers = pd.DataFrame({
    'seller_id': [f'SELL{str(i).zfill(5)}' for i in range(1, N_SELLERS+1)],
    'seller_name': [fake.company() for _ in range(N_SELLERS)],
    'city': np.random.choice(SELLER_CITIES, size=N_SELLERS),
    'rating': seller_ratings,
    'total_products': np.random.randint(5, 500, N_SELLERS),
    'fulfilment_type': np.random.choice(
        ['Flipkart Fulfilled', 'Seller Fulfilled'],
        size=N_SELLERS,
        p=[0.40, 0.60]
    ),
    'join_date': [
        (datetime(2024,1,1) - timedelta(days=np.random.randint(30, 1825))).strftime('%Y-%m-%d')
        for _ in range(N_SELLERS)
    ]
})

sellers.to_csv('data/sellers.csv', index=False)
print(f"✓ Sellers: {len(sellers):,} rows")
print(f"Avg seller rating: {sellers['rating'].mean():.2f}")
print(sellers['fulfilment_type'].value_counts())
```

```python
N_PRODUCTS = 20000

CATEGORIES = {
    'Electronics':   {'base_return_rate': 0.18, 'subcats': ['Smartphones','Laptops','Tablets','Headphones','Cameras']},
    'Fashion':       {'base_return_rate': 0.22, 'subcats': ['Mens Clothing','Womens Clothing','Footwear','Accessories']},
    'Home & Kitchen':{'base_return_rate': 0.10, 'subcats': ['Cookware','Furniture','Decor','Appliances']},
    'Books':         {'base_return_rate': 0.03, 'subcats': ['Fiction','Non-Fiction','Academic','Comics']},
    'Sports':        {'base_return_rate': 0.12, 'subcats': ['Fitness','Outdoor','Team Sports','Yoga']},
    'Beauty':        {'base_return_rate': 0.08, 'subcats': ['Skincare','Haircare','Makeup','Fragrances']},
    'Toys':          {'base_return_rate': 0.14, 'subcats': ['Action Figures','Board Games','Educational','Outdoor Toys']},
    'Grocery':       {'base_return_rate': 0.02, 'subcats': ['Snacks','Beverages','Health Foods','Staples']},
}

categories = np.random.choice(list(CATEGORIES.keys()), size=N_PRODUCTS,
    p=[0.25, 0.22, 0.15, 0.08, 0.10, 0.08, 0.07, 0.05])

subcategories = [
    np.random.choice(CATEGORIES[c]['subcats']) for c in categories
]

base_return_rates = np.array([CATEGORIES[c]['base_return_rate'] for c in categories])

# Price ranges by category
price_ranges = {
    'Electronics': (5000, 150000), 'Fashion': (299, 8000),
    'Home & Kitchen': (499, 25000), 'Books': (99, 999),
    'Sports': (299, 15000), 'Beauty': (199, 5000),
    'Toys': (199, 5000), 'Grocery': (49, 999)
}

prices = np.array([
    round(np.random.uniform(*price_ranges[c]), 2)
    for c in categories
])

image_counts = np.random.randint(1, 10, N_PRODUCTS)
avg_ratings = np.round(np.clip(np.random.normal(3.9, 0.6, N_PRODUCTS), 1.0, 5.0), 1)
seller_ids = np.random.choice(sellers['seller_id'], size=N_PRODUCTS)

products = pd.DataFrame({
    'product_id':       [f'PROD{str(i).zfill(6)}' for i in range(1, N_PRODUCTS+1)],
    'seller_id':        seller_ids,
    'product_name':     [f"{cat} Product {fake.word().title()}" for cat in categories],
    'category':         categories,
    'subcategory':      subcategories,
    'price':            prices,
    'avg_rating':       avg_ratings,
    'image_count':      image_counts,
    'base_return_rate': np.round(base_return_rates, 3),
})

products.to_csv('data/products.csv', index=False)
print(f"✓ Products: {len(products):,} rows")
print(products['category'].value_counts())
```

```python
N_ORDERS = 100000

order_dates = [
    datetime(2024, 1, 1) + timedelta(days=np.random.randint(0, 365))
    for _ in range(N_ORDERS)
]

payment_methods = np.random.choice(
    ['COD', 'Credit Card', 'Debit Card', 'UPI', 'Net Banking', 'Wallet'],
    size=N_ORDERS,
    p=[0.30, 0.18, 0.15, 0.25, 0.07, 0.05]
)

# Delivery delay — embedded signal
promised_days = np.random.randint(2, 7, N_ORDERS)
# Some sellers cause delays
actual_extra = np.random.choice([0,0,0,1,2,3,5,7], size=N_ORDERS, p=[0.4,0.2,0.1,0.1,0.08,0.06,0.04,0.02])
actual_days = promised_days + actual_extra

orders = pd.DataFrame({
    'order_id': [f'ORD{str(i).zfill(7)}' for i in range(1, N_ORDERS+1)],
    'customer_id': np.random.choice(customers['customer_id'], size=N_ORDERS),
    'order_date': [d.strftime('%Y-%m-%d') for d in order_dates],
    'payment_method': payment_methods,
    'promised_delivery_days': promised_days,
    'actual_delivery_days': actual_days,
    'delivery_delay_days': actual_extra,
    'order_value': 0.0  # will fill after items
})

orders.to_csv('data/orders.csv', index=False)
print(f"✓ Orders: {len(orders):,} rows")
print(pd.Series(payment_methods).value_counts())
```

```python
print("Generating order items with embedded return signals...")

order_items_list = []
N_ITEMS = 200000

# Randomly assign products to orders
sampled_orders = np.random.choice(orders['order_id'], size=N_ITEMS)
sampled_products = np.random.choice(range(len(products)), size=N_ITEMS)

prod_df = products.set_index('product_id')
sel_df = sellers.set_index('seller_id')
cust_df = customers.set_index('customer_id')
ord_df = orders.set_index('order_id')

RETURN_REASONS = [
    'Product damaged', 'Wrong item delivered', 'Size/fit issue',
    'Product not as described', 'Quality issue', 'Changed mind',
    'Defective product', 'Late delivery', 'Better price elsewhere'
]

items = []
returns_list = []
return_counter = 1

for i in range(N_ITEMS):
    oid = sampled_orders[i]
    pidx = sampled_products[i]
    prod_row = products.iloc[pidx]
    pid = prod_row['product_id']
    sid = prod_row['seller_id']
    
    ord_row = ord_df.loc[oid]
    cust_id = ord_row['customer_id']
    cust_row = cust_df.loc[cust_id]
    sel_row = sel_df.loc[sid]
    
    qty = np.random.choice([1,1,1,2,3], p=[0.7,0.15,0.08,0.05,0.02])
    discount = round(np.random.choice([0,5,10,15,20,30,40,50,60],
                     p=[0.3,0.1,0.15,0.1,0.1,0.08,0.07,0.06,0.04]) / 100, 2)
    unit_price = prod_row['price']
    final_price = round(unit_price * (1 - discount) * qty, 2)
    
    # ── RETURN PROBABILITY SIGNAL ──────────────────────────
    p = prod_row['base_return_rate']
    
    # Seller rating effect
    if sel_row['rating'] < 3.0:   p *= 2.2
    elif sel_row['rating'] < 3.5: p *= 1.6
    elif sel_row['rating'] > 4.5: p *= 0.7
    
    # Payment method effect
    if ord_row['payment_method'] == 'COD': p *= 1.35
    
    # Delivery delay effect
    delay = ord_row['delivery_delay_days']
    if delay >= 5:   p *= 1.8
    elif delay >= 3: p *= 1.4
    elif delay >= 1: p *= 1.15
    
    # High discount effect (impulse buy)
    if discount >= 0.50: p *= 1.5
    elif discount >= 0.40: p *= 1.3
    
    # New customer effect
    if cust_row['account_age_days'] < 30: p *= 1.4
    
    # Low image count
    if prod_row['image_count'] <= 2: p *= 1.25
    
    # High price effect
    if unit_price > 50000: p *= 1.3
    elif unit_price > 20000: p *= 1.15
    
    # Low product rating
    if prod_row['avg_rating'] < 3.0: p *= 1.5
    elif prod_row['avg_rating'] < 3.5: p *= 1.2
    
    # Flipkart fulfilled = lower return
    if sel_row['fulfilment_type'] == 'Flipkart Fulfilled': p *= 0.85
    
    p = min(p, 0.75)  # cap at 75%
    
    is_returned = int(np.random.random() < p)
    
    item_id = f'ITEM{str(i+1).zfill(8)}'
    items.append({
        'item_id': item_id,
        'order_id': oid,
        'product_id': pid,
        'seller_id': sid,
        'quantity': qty,
        'unit_price': unit_price,
        'discount_pct': discount * 100,
        'final_price': final_price,
        'is_returned': is_returned
    })
    
    if is_returned:
        return_date = (datetime.strptime(str(ord_row['order_date']), '%Y-%m-%d')
                       + timedelta(days=np.random.randint(1, 15)))
        
        # Reason weighted by category
        cat = prod_row['category']
        if cat == 'Fashion':
            reason = np.random.choice(RETURN_REASONS,
                p=[0.05,0.05,0.35,0.20,0.10,0.10,0.05,0.05,0.05])
        elif cat == 'Electronics':
            reason = np.random.choice(RETURN_REASONS,
                p=[0.20,0.10,0.02,0.20,0.25,0.05,0.12,0.04,0.02])
        else:
            reason = np.random.choice(RETURN_REASONS)
        
        cost_per_return = round(final_price * np.random.uniform(0.10, 0.25), 2)
        
        returns_list.append({
            'return_id': f'RET{str(return_counter).zfill(7)}',
            'item_id': item_id,
            'order_id': oid,
            'product_id': pid,
            'seller_id': sid,
            'return_date': return_date.strftime('%Y-%m-%d'),
            'return_reason': reason,
            'return_cost': cost_per_return,
            'resolution': np.random.choice(
                ['Refund', 'Replacement', 'Store Credit'],
                p=[0.65, 0.25, 0.10]
            )
        })
        return_counter += 1
    
    if (i+1) % 50000 == 0:
        print(f"  {i+1:,} / {N_ITEMS:,} items processed...")

order_items = pd.DataFrame(items)
returns_df = pd.DataFrame(returns_list)

order_items.to_csv('data/order_items.csv', index=False)
returns_df.to_csv('data/returns.csv', index=False)

print(f"\n✓ Order items: {len(order_items):,} rows")
print(f"✓ Returns: {len(returns_df):,} rows")
print(f"✓ Overall return rate: {order_items['is_returned'].mean()*100:.2f}%")
```

```python
# Update order value from items
order_totals = order_items.groupby('order_id')['final_price'].sum().reset_index()
order_totals.columns = ['order_id', 'order_value']
orders = orders.drop('order_value', axis=1).merge(order_totals, on='order_id', how='left')
orders['order_value'] = orders['order_value'].fillna(0).round(2)
orders.to_csv('data/orders.csv', index=False)

# Final validation
print("=" * 50)
print("DATASET GENERATION COMPLETE")
print("=" * 50)
for f in ['customers', 'sellers', 'products', 'orders', 'order_items', 'returns']:
    df = pd.read_csv(f'data/{f}.csv')
    print(f"  {f:<15}: {len(df):>8,} rows | {len(df.columns)} columns")

print()
print("Return rate by category:")
merged = order_items.merge(products[['product_id','category']], on='product_id')
print(merged.groupby('category')['is_returned'].mean().mul(100).round(2).sort_values(ascending=False).to_string())
print()
print(f"Total estimated return cost: ₹{returns_df['return_cost'].sum():,.0f}")
```
